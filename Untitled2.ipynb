```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt

# Превращаем картинки в тензоры и нормализуем пиксели
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])

# Скачиваем обучающую и тестовую выборки Fashion-MNIST
trainset = torchvision.datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)

testset = torchvision.datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

# Названия классов одежды
classes = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

print("Датасет успешно загружен! Количество батчей для обучения:", len(trainloader))
```

```python
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN, self).__init__()
        # Сверточные слои (ищут узоры на картинках)
        self.features = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2), # Уменьшаем картинку в 2 раза
            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2)
        )
        # Полносвязные слои (принимают решение на основе признаков)
        self.classifier = nn.Sequential(
            nn.Linear(32 * 7 * 7, 128),
            nn.ReLU(),
            nn.Linear(128, 10) # На выходе 10 классов одежды
        )

    def forward(self, x):
        x = self.features(x)
        x = x.view(x.size(0), -1) # Вытягиваем в один вектор
        x = self.classifier(x)
        return x

net = SimpleCNN()
print(net)
```

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters(), lr=0.001)

print("Начинаем обучение нейросети...")
for epoch in range(3):  # 3 эпохи для быстрого результата
    running_loss = 0.0
    for i, data in enumerate(trainloader, 0):
        inputs, labels = data

        optimizer.zero_grad()       # Очищаем старые градиенты
        outputs = net(inputs)       # Предсказание модели
        loss = criterion(outputs, labels) # Считаем ошибку
        loss.backward()             # Обратное распространение ошибки
        optimizer.step()            # Обновляем веса

        running_loss += loss.item()

    print(f'Эпоха [{epoch + 1}/3] завершена. Потери (Loss): {running_loss / len(trainloader):.3f}')

print('Обучение закончено!')
```

```python
correct = 0
total = 0

with torch.no_grad():
    for data in testloader:
        images, labels = data
        outputs = net(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total
print(f'Точность нашей нейросети на тестовых изображениях: {accuracy:.2f}%')
```
